```python
import pickle
import torch
import io
class CPU_Unpickler(pickle.Unpickler):
    def find_class(self, module, name):
        if module == 'torch.storage' and name == '_load_from_bytes':
            return lambda b: torch.load(io.BytesIO(b), map_location='cpu')
        else:
            return super().find_class(module, name)
```

```python
import pickle
layer_head_A = []
for i in range(number_of_examples):
    with open(f'/home/t-josingh/How-To-Think-Step-by-Step/InformationFlow/result/5/A/example_{i}.pickle', 'rb') as f:
        data = pickle.load(f)
    # if(i < 10):
    #     with open('/home/t-josingh/How-To-Think-Step-by-Step/InformationFlow/result/reasoning/combined/5/A/accuracy/Activation_example_start_0_end_10.pickle', 'rb') as f:
    #         input_ids = CPU_Unpickler(f).load()
    #     input_ids = input_ids['input_ids']
    layer_head_pair = []
    for d in data:
        layer_head_pair.append((d['layer'], d['head']))
        for attending_layer_head in d['attending_layer_head']:
            layer_head_pair.append((attending_layer_head['layer'], attending_layer_head['head']))
    layer_head_pair = list(set(layer_head_pair))
    layer_head_pair = sorted(layer_head_pair, key=lambda x: x[0])
    layer_head_A.append(layer_head_pair)
```

```python
import pickle
layer_head_B = []
for i in range(number_of_examples):
    with open(f'/home/t-josingh/How-To-Think-Step-by-Step/InformationFlow/result/5/B/example_{i}.pickle', 'rb') as f:
        data = pickle.load(f)
    layer_head_pair = []
    for d in data:
        layer_head_pair.append((d['layer'], d['head']))
        for attending_layer_head in d['attending_layer_head']:
            layer_head_pair.append((attending_layer_head['layer'], attending_layer_head['head']))
    layer_head_pair = list(set(layer_head_pair))
    layer_head_pair = sorted(layer_head_pair, key=lambda x: x[0])
    layer_head_B.append(layer_head_pair)
```

```python
import pickle
layer_head_C = []
for i in range(number_of_examples):
    with open(f'/home/t-josingh/How-To-Think-Step-by-Step/InformationFlow/result/5/C/example_{i}.pickle', 'rb') as f:
        data = pickle.load(f)
    layer_head_pair = []
    for d in data:
        layer_head_pair.append((d['layer'], d['head']))
        for attending_layer_head in d['attending_layer_head']:
            layer_head_pair.append((attending_layer_head['layer'], attending_layer_head['head']))
    layer_head_pair = list(set(layer_head_pair))
    layer_head_pair = sorted(layer_head_pair, key=lambda x: x[0])
    layer_head_C.append(layer_head_pair)
```

```python
combined_A = []
for i in range(number_of_examples):
    combined_A += layer_head_A[i]
combined_A = set(combined_A)

combined_B = []
for i in range(number_of_examples):
    combined_B += layer_head_B[i]
combined_B = set(combined_B)

combined_C = []
for i in range(number_of_examples):
    combined_C += layer_head_C[i]
combined_C = set(combined_C)
```
